**json_data2 represents the following:**

A structured catalog of digital colors designed for color theory analysis and UI design applications.

**Columns**

**1. Name**
* lists the name of different colors

**2. Hex**
* lists the exact hexadecimal value that provides a standardized, universally recognized code for computer systems, web browsers, and design softwares to accurately render the specific color on a screen

**3. Category**
* classifies the color based on its color theory placement or temperature profile to allow designers or applications to easily group, filter, and query colors by their functional role in a palette

**4, 5 and 6. Rgb.r/ Rgb.g / Rgb.b**
* classifies the individual intensities of the red, green, and blue channels that make up the RGB color model. The primary purpose of these three combined features is to break down the color into its fundamental light components, allowing digital screens and graphics software to blend them together to reproduce the exact color.
